## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*



I am going with **Lane 2 — Refresh / Content Opportunity Scoring**. I want to find which content items may be worth reviewing first. The dataset has useful information about traffic, clicks, engagement, content age, and recent performance, so this lane seems practical to explore further.

**CODE CELL**

```python
import pandas as pd

DATA_PATH = "../../data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]:,}")
print(f"Unique clients: {df['client_id'].nunique():,}")
print(f"Unique content items: {df['content_id'].nunique():,}")
```

## 2. The question: decision, action, cost of a wrong call

The main question is: **Which content items should be reviewed first?**

The SEO/content team could use the ranking to decide which pages to check and whether they need a refresh, expansion, protection, pruning, or monitoring.

A wrong recommendation could waste the team's review time. Missing an important page could also mean missing a useful content opportunity.

**Unit of analysis:** one content item.

**CODE CELL**

```python
trend_counts = df["trend_direction"].value_counts(dropna=False)
trend_pct = (
    df["trend_direction"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(1)
)

print("Trend direction counts:")
print(trend_counts)

print("\nTrend direction percentages:")
print(trend_pct)
```

## 3. Quick look at the data (2-3 real numbers)

The starter dataset has **30,000 content items from 32 clients**. I found that **16,262 items (54.2%) have a downward trend** in the current data. This gives enough variation to make content prioritization worth exploring.

**CODE CELL**

```python
# Real numbers used in this section

total_items = df["content_id"].nunique()
total_clients = df["client_id"].nunique()

down_count = (df["trend_direction"] == "down").sum()
down_pct = (down_count / len(df)) * 100

print(f"Unique content items: {total_items:,}")
print(f"Unique clients: {total_clients:,}")
print(f"Items with trend_direction='down': {down_count:,} ({down_pct:.1f}%)")
```

It's Optional

```python
last30_impressions = df["impressions_last_30d"].sum()
prev30_impressions = df["impressions_prev_30d"].sum()

last30_clicks = df["clicks_last_30d"].sum()
prev30_clicks = df["clicks_prev_30d"].sum()

print(f"Last-30-day impressions: {last30_impressions:,}")
print(f"Previous-30-day impressions: {prev30_impressions:,}")
print(f"Last-30-day clicks: {last30_clicks:,}")
print(f"Previous-30-day clicks: {prev30_clicks:,}")
```
